# Helper functions to calculate fields in ArcPro after running Process EIA

### Before beginning
- your "Landscape Context Comments" and "Condition Rating Comments" fields should be blank/null. See transfer field function if needed to append these text to "Landscape Context" and "Condition of EO" fields

#### function to add the EIA summarized field notes into Landscape or Condition of EO

```python
# example usage
# summarize_fields(!L0CNHP_All_Field_Observations_2026.landscape_context!, !test2.landscape_context_comm!)

def summarize_fields(fieldA, fieldB):
    if fieldA in (None, ".", ""):
        if fieldB not in (None, "None"):
            return str(fieldB)
        else:
            return None
    elif fieldB not in (None, "None"):
        return (str(fieldA) + " " + str(fieldB))
    else:
        return str(fieldA)


```

#### function to overwrite Landscape Rank and Condition Rank that were assigned in the field (usually were arbitrarily A)

**function to add EIA calculated EO rank to Overall EO Rank Comments**

```python
# example usage
# summarize_fields(!L0CNHP_All_Field_Observations_2026.eo_rank_com!, !test2.eo_rank!)

def summarize_fields(fieldA, fieldB):
    if fieldA in (None, ".", ""):
        if fieldB not in (None, "None"):
            return ("EIA calculated rank: " + str(fieldB) + ".")
        else:
            return None
    elif fieldB not in (None, "None"):
        return ("EIA calculated rank: " + str(fieldB) + ". " + str(fieldA))
    else:
        return str(fieldA)
```
